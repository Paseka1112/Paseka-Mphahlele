```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

college = pd.read_csv('College.csv')
college.head(10)
```

```python
college2 = pd.read_csv('College.csv', index_col= 0)
college3 = college.rename({'Unnamed: 0': 'College'}, axis = 1)
college3 = college3.set_index('College')
college3
```

```python
college = college3
college.describe()
```

```python
pd.plotting.scatter_matrix(college[['Top10perc', 'Apps', 'Enroll']]);
```

```python
college.boxplot(column= "Outstate", by = "Private")
plt.title('Outstate vs Private')
```

```python
college['Elite'] = pd.cut(college['Top10perc'],[0,0.5,1], labels =['No', 'Yes'])
pd.Series(college['Elite']).value_counts()
```

```python
college.boxplot(column= "Outstate", by = "Elite")
plt.title('Outstate vs Elite')
```

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

college['Outstate'].plot.hist(ax=axes[0, 0], bins=10)
axes[0, 0].set_title('Outstate (bins=10)')

college['Enroll'].plot.hist(ax=axes[0, 1], bins=30)
axes[0, 1].set_title('Enroll (bins=30)')

college['Top10perc'].plot.hist(ax=axes[1, 0], bins=15)
axes[1, 0].set_title('Top10perc (bins=15)')

college['PhD'].plot.hist(ax=axes[1, 1], bins=20)
axes[1, 1].set_title('PhD (bins=20)')

plt.tight_layout()
plt.show()
```
